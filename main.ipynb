```python
paragraph = """ Khan Abdul Ghaffar Khan[a] (6 February 1890 – 20 January 1988),
also known as Bacha Khan[b] and Badshah Khan,
[c] was an Indian independence activist from the North-West Frontier Province,
 and founder of the Khudai Khidmatgar resistance movement against British rule in colonial India.[2]
  After the partition occurred, he became a Pakistani politician and led the Azad Party.[3]

He was a political and spiritual leader known for his nonviolent opposition and lifelong pacifism;
 he was a devout Muslim and an advocate for Hindu–Muslim unity in the subcontinent.
 [4] Because of his similar ideology and close friendship with Mahatma Gandhi, Khan was nicknamed "the Frontier Gandhi" (سرحدی گاندھی, Sarhadi Gandhi).
 [5][6] In 1929, Khan founded the Khudai Khidmatgar, an anti-colonial nonviolent resistance movement.
 [7] The Khudai Khidmatgar's success and popularity eventually prompted the colonial government to launch numerous crackdowns against Khan and his supporters; 
 the Khudai Khidmatgar experienced some of the most severe repression of the entire Indian independence movement.[8]
"""
```

```python
print("\n:\n", paragraph)
```

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

sentences = nltk.sent_tokenize(paragraph)

print(sentences)
```

```python
type(sentences)
```

```python
stemmer=PorterStemmer()
```

```python
stemmer.stem("history")
```

```python
# apply lemmatization 
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("history")
```

```python
len(sentences)
```

```python
# now clean special characters, numbers, and symbols from the sentences and keep only alphabets. Also, convert all the sentences to lower case.
# by using regular expressions (re)
import re
corpus = []
for i in range(len(sentences)):
    review = re.sub('[^a-zA-Z]', ' ', sentences[i])
    review = review.lower()
    review = review.split()
    review = [lemmatizer.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
stopwords.words('english')
```

```python
#stemming 
for i in corpus:
    # Apply stemming to each word in the sentence and print the stemmed words
    word = nltk.word_tokenize(i)
    for word in word:
        if word not in set(stopwords.words('english')):
            print(stemmer.stem(word))
```

```python
# for lemmatization 
for i in corpus:
    # Apply lemmatization to each word in the sentence and print the lemmatized words
    word = nltk.word_tokenize(i)
    for word in word:
        if word not in set(stopwords.words('english')):
            print(lemmatizer.lemmatize(word))
```

```python
# now BAG OF WORDS MODEL
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
X=cv.fit_transform(corpus)
```
